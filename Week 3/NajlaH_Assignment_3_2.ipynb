```python
import pandas as pd
url = "https://www.canada.ca/en/public-health/services/surveillance/respiratory-virus-detections-canada/2018-2019/respiratory-virus-detections-isolations-week-01-ending-january-5-2019.html"
tables = pd.read_html(url)
len(tables)

for i, table in enumerate(tables):
    print(i, table.shape)
    print(table.head(2))
    print()

```

```python
table2 = tables[1].copy()
table2.head()
table2['Reporting Laboratory'] = (
    table2['Reporting Laboratory']
    .astype(str)
    .str.strip()
)
provinces_territories = [
    'Newfoundland',
    'Prince Edward Island',
    'Nova Scotia',
    'New Brunswick',
    'Province of Québec',
    'Province of Ontario',
    'Manitoba',
    'Province of Saskatchewan',
    'Province of Alberta',
    'British Columbia',
    'Yukon',
    'Northwest Territories',
    'Nunavut'
]

province_data = table2[
    table2['Reporting Laboratory'].isin(provinces_territories)
].copy()
```

```python
province_data = province_data.rename(
    columns={'Reporting Laboratory': 'Province/Territory'}
)

province_data
```
